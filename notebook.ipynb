```python
import pandas as pd
import pathlib
from nba_api.stats.endpoints import leaguegamefinder
```

```python
games = leaguegamefinder.LeagueGameFinder(
    season_nullable="2024-25",
    season_type_nullable="Regular Season"
)
df = games.league_game_finder_results.get_data_frame()
```

```python
# Count the number of @ games
n_at_games = len(df[df["MATCHUP"].str.contains("@")])
# Count the number of vs. games
n_vs_games = len(df[df["MATCHUP"].str.contains("vs.", regex=False)])

print(f"Number of @ games: {n_at_games}")
print(f"Number of vs. games: {n_vs_games}")
```

```python
# Create column that returns True if the team is the home team
df["is_home"] = df["MATCHUP"].str.contains("vs.", regex=False)

# Group by GAME_ID and sum "is_home"; filter for rows where sum is 0
home_counts = df.groupby("GAME_ID")["is_home"].sum()

df[df["GAME_ID"].isin(home_counts[home_counts==0].index)]
```

There are five games that are included in the regular season, but not played in the home stadiums:
- 2025-01-25, IND against SAS, in Paris
- 2025-01-23, IND against SAS, in Paris
- 2024-12-14, HOU against OKC, in Las Vegas
- 2024-12-14, ATL against MIL, in Las Vegas
- 2024-11-02, MIA against WAS, in Mexico
